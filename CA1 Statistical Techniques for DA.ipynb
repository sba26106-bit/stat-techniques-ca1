### 1.2 Removing inconsistent data

```python
# Drop columns that contain only NA values
clean_df = df.dropna(axis=1, how="all")

# Drop [Score] columns where every value is '-- / 0'
score_cols = [c for c in clean_df.columns if c.endswith("[Score]")]

all_blank = clean_df[score_cols].apply(lambda s: s.str.strip().eq("-- / 0").all())

clean_df = clean_df.drop(columns=all_blank[all_blank].index)

# Remove 'Total score' column
clean_df = clean_df.drop(columns="Total score")

# Remove 'Unnamed: 0' column
clean_df = clean_df.drop(columns="Unnamed: 0")

# Strip spaces
text_cols = clean_df.select_dtypes(include=['object', 'string']).columns
clean_df[text_cols] = clean_df[text_cols].apply(lambda s: s.str.strip())

# Rename columns to short names
clean_df.columns = [
    "gender", "age", "years_in_ireland", "county", "living_area",
    "household", "has_car", "holiday_pref", "trips_total", "trip_type",
    "travel_with", "trip_purpose", "trips_ireland", "destination_type",
    "transport", "trip_length", "counties_visited", "satisfaction",
]
```

Some questions in the Google Form had an "Other:" option, so some answers were typed in by hand. These answers need to be processed.

```python
# Process hand-typed answers

# the fixed options for each question that had an "Other" box
form_options = {
    'living_area':
        ['Large city', 'Small / Medium-sized town', 'Village', 'Rural area'],
    'trip_purpose':
        ['Exploring new places', 'Relaxation', 'Visiting friends and family'],
    'destination_type':
        ['A combination of these', 'Coastal areas and beaches',
         'Mountains and countryside', 'National parks and nature reserves',
         'Cities and towns', 'Historical sites and castles'],
}

def find_hand_typed(df, form_options):
    """Return every answer that is not one of the form's fixed options."""
    found = []
    for question, options in form_options.items():
        answers = df[question].dropna().str.strip()
        typed = answers[~answers.isin(options)]
        for row, answer in typed.items():
            found.append({'row': row, 'question': question, 'answer': answer})
    return pd.DataFrame(found, columns=['row', 'question', 'answer'])

hand_typed = find_hand_typed(clean_df, form_options)
hand_typed
```

```python
# Replace hand-typed answers
recode = {
    'living_area': {
        'Dublin (small city)': 'Large city',
        'Small city, Dublin':  'Large city',
    },
    'trip_purpose': {
        'All of the above': np.nan,
    },
    'destination_type': {
        'Abroad': np.nan,
    },
}

clean_df = clean_df.replace(recode)

# Check if hand-typed values still exist
find_hand_typed(clean_df, form_options)
```

Some respondents listed their county of residence among the counties they visited. Because the analysis is about travelling around Ireland rather than within the home county, the home county is removed from each respondent's list.

```python
HOME_COL = 'county'                # Q4 - county of residence
VISITED_COL = 'counties_visited'   # Q17 - counties visited

def split_counties(answer):
    """Return the counties in one Q17 answer as a list."""
    if pd.isna(answer):
        return []
    return [c.strip() for c in answer.split(';') if c.strip() != '']

def count_listed_home(df):
    """Return how many respondents listed their home county as visited."""
    listed_home = df.apply(
        lambda row: row[HOME_COL] in split_counties(row[VISITED_COL]),
        axis=1)
    return listed_home.sum()

# Check how many respondents listed their own county as a visited county
print('Respondents who listed their home county:', count_listed_home(clean_df))
```

32 people listed their home county and 9 did not, so removing it makes the responses comparable.

```python
def remove_home_county(row):
    """Return the Q17 answer without the respondent's own county."""
    kept = [c for c in split_counties(row[VISITED_COL]) if c != row[HOME_COL]]
    return ';'.join(kept)

# Remove the county of residence from the list of counties visited
clean_df[VISITED_COL] = clean_df.apply(remove_home_county, axis=1)

# Check again
print('Respondents who listed their home county:', count_listed_home(clean_df))
```

Check the dataset for duplicated rows, which may be repeat submissions from the same person.

Two questions in the Google Form are related and may contradict each other. Q9 asks for the total number of trips (in Ireland and abroad) and Q13 asks for the number of trips within Ireland. If a respondent answered incorrectly, the second number can be larger than the first. Such rows are flagged in an additional column. The respondents' other answers are still used, and each of the two trip counts is still analysed on its own. The flagged rows are excluded only from analyses that use both counts together, such as comparing them or calculating the share of trips taken in Ireland.

```python
# Count respondents who reported more trips within Ireland than trips in total
inconsistent = clean_df['trips_ireland'] > clean_df['trips_total']

print("Inconsistent rows:", inconsistent.sum(), "of", len(clean_df))
```

```python
# Add a column flagging inconsistent trip counts
clean_df["trips_inconsistent"] = inconsistent
```

Add a column with the number of counties visited for leisure.

```python
# Number of counties visited for leisure (Q17 is multi-select, separated by ';')
def count_counties(answer):
    """Return how many counties are listed in one Q17 answer."""
    return len(split_counties(answer))

clean_df['n_counties'] = clean_df[VISITED_COL].apply(count_counties)
```

Add two columns with the maximum and average travel distance within Ireland, based on the distances between county towns. Because these values do not take into account the exact places of residence and travel, they are only approximate. The distances are calculated from the latitude and longitude of each county town.

```python
# Add columns with maximum and average distance of travelling inside Ireland

# Latitude and longitude of each county town (decimal degrees)
county_towns = {
    'Carlow':    (52.836, -6.926),   # Carlow
    'Cavan':     (53.991, -7.360),   # Cavan
    'Clare':     (52.843, -8.986),   # Ennis
    'Cork':      (51.898, -8.476),   # Cork
    'Donegal':   (54.833, -7.483),   # Lifford
    'Dublin':    (53.350, -6.260),   # Dublin
    'Galway':    (53.271, -9.049),   # Galway
    'Kerry':     (52.271, -9.700),   # Tralee
    'Kildare':   (53.216, -6.667),   # Naas
    'Kilkenny':  (52.654, -7.245),   # Kilkenny
    'Laois':     (53.033, -7.300),   # Portlaoise
    'Leitrim':   (53.945, -8.087),   # Carrick-on-Shannon
    'Limerick':  (52.664, -8.623),   # Limerick
    'Longford':  (53.727, -7.800),   # Longford
    'Louth':     (54.004, -6.404),   # Dundalk
    'Mayo':      (53.855, -9.298),   # Castlebar
    'Meath':     (53.653, -6.681),   # Navan
    'Monaghan':  (54.249, -6.968),   # Monaghan
    'Offaly':    (53.275, -7.493),   # Tullamore
    'Roscommon': (53.627, -8.189),   # Roscommon
    'Sligo':     (54.272, -8.476),   # Sligo
    'Tipperary': (52.862, -8.200),   # Nenagh
    'Waterford': (52.259, -7.110),   # Waterford
    'Westmeath': (53.526, -7.338),   # Mullingar
    'Wexford':   (52.336, -6.463),   # Wexford
    'Wicklow':   (52.980, -6.044),   # Wicklow
}

# Function to calculate the distance between two county towns from their latitude and longitude
def haversine_km(county_a, county_b):
    """Straight-line (great-circle) distance in km between two county towns."""
    lat1, lon1 = np.radians(county_towns[county_a])
    lat2, lon2 = np.radians(county_towns[county_b])
    a = (np.sin((lat2 - lat1) / 2) ** 2
         + np.cos(lat1) * np.cos(lat2) * np.sin((lon2 - lon1) / 2) ** 2)
    return 2 * 6371 * np.arcsin(np.sqrt(a))   # 6371 km = mean radius of the Earth

# 26 x 26 table of distances between all county towns
counties = list(county_towns)
dist_matrix = pd.DataFrame(
    [[haversine_km(a, b) for b in counties] for a in counties],
    index=counties, columns=counties).round(1)

dist_matrix.head()
```

```python
# Function to get the average and maximum distance for each respondent
def trip_distances(row):
    """Average and maximum distance from the home county to the visited counties."""
    visited = split_counties(row[VISITED_COL])
    if len(visited) == 0:
        return pd.Series([np.nan, np.nan])
    distances = dist_matrix.loc[row[HOME_COL], visited]
    return pd.Series([distances.mean(), distances.max()])

# Add the new columns
clean_df[['avg_distance_km', 'max_distance_km']] = clean_df.apply(trip_distances, axis=1).round(1)

clean_df.head()
```

Add a column with the share of leisure trips taken within Ireland (trips in Ireland divided by total trips). The share is not calculated for rows flagged as inconsistent, or where the total number of trips is zero.

```python
# Check values
valid = ~clean_df['trips_inconsistent'] & (clean_df['trips_total'] > 0)

# Share of leisure trips taken within Ireland (0 = all abroad, 1 = all in Ireland)
clean_df['share_ireland'] = (
    clean_df['trips_ireland'] / clean_df['trips_total']
).where(valid).round(2)

# Quick check
print('Rows with a share:', clean_df['share_ireland'].notna().sum(), 'of', len(clean_df))
clean_df.head(10)
```
